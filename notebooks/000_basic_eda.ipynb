### Constant Columns

```python
constant_cols = []
cardinality = df.nunique()
constant_cols = cardinality[cardinality == 1].index.tolist()
if constant_cols:
    for col in constant_cols:
        constant_value = df[col].iloc[0] 
        print(f"'{col}': '{constant_value}'")
else:
    print("No constant columns detected")

```

### Cardinality

```python
cardinality = df.nunique().sort_values(ascending=False)
print(cardinality.to_frame('Unique Values'))
```

- nameOrig: Extremely high → probably a unique identifier for each sender (account ID). Drop or use for - aggregation
- amount: Continuous numeric feature. Maybe scale
- oldbalanceDest: Continuous numeric, possibly many zeros? Keep it
- newbalanceDest: Continuous numeric. Keep
- nameDest: High-cardinality categorical. Too many uniques → Drop, hash, or aggregate
- newbalanceOrig: Continuous numeric. Keep
- oldbalanceOrig: Continuous numeric. Keep
- step: Time step. Keep, can encode as time (sin/cos transformation or directly numeric)
- type: Transaction type. Keep OHT
- isFlaggedFraud: Binary target. Keep ofc

```python
high_cardinality_cols = []
for col in df.select_dtypes(include='object').columns:
        if df[col].nunique() / len(df) > 0.95:
            high_cardinality_cols.append(f"'{col}' has very high cardinality ({df[col].nunique():,} unique values)")

for col in high_cardinality_cols:
    print(col)
print("-" *50)
print(f" The length of the dataframe is: {len(df):,}")
```

### Negative values 

```python
neg_vals = []
for col in df.select_dtypes(include=[np.number]).columns:
    if (df[col] < 0).any():
        n_negative = (df[col] < 0).sum()
        neg_vals.append(f" This {col} has {n_negative:,} negative values")

if neg_vals:
    for val in neg_vals:
        print(neg_vals)
else:
    print("No obvious negative values detected")
```

### Outliers

```python
for col in df.select_dtypes(include=[np.number]).columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    if len(outliers) > 0:
        pct = len(outliers) / len(df) * 100
        print(f"  {col}: {len(outliers):,} outliers ({pct:.2f}%)")
```

- step : probably not a real outlier
- amount: expected, transactions are usually heavy tailed
- old/new balances in Orig/Dest: also expected. Balances vary a lot across users
- isFlaggedFraud: the target

## Business consistency

```python
data = df.copy()

inconsistent_cols = []
data['balance_diff'] = data['oldbalanceOrg'] - data['newbalanceOrig']

# check balance_diff and amount are equal, with allowed difference of |1|
inconsistent = data[~np.isclose(data['balance_diff'], data['amount'], atol=1.0)]
if len(inconsistent) > 0:
    pct = len(inconsistent) / len(data) * 100
    inconsistent_cols.append(f" {len(inconsistent):,} transactions ({pct:.2f}%) have inconsistent balance calculations")
    print(inconsistent_cols)
else:
    print(" Balance calculations are consistent")
```

```python
data['is_inconsistent_origin'] = ~np.isclose(data['oldbalanceOrg'] - data['newbalanceOrig'], data['amount'], atol=1.0)
data['is_inconsistent_dest'] = ~np.isclose(data['newbalanceDest'] - data['oldbalanceDest'], data['amount'], atol=1.0)

summary = (
    data.groupby('type')[['is_inconsistent_origin', 'is_inconsistent_dest']]
      .mean()
      .mul(100)
      .rename(columns={
          'is_inconsistent_origin': '%_inconsistent_origin',
          'is_inconsistent_dest': '%_inconsistent_dest'
      })
      .sort_values('%_inconsistent_origin', ascending=False)
)
summary
```

There are many inconsistencies but they should be analyzed separatedly because some type of movements only affect one either Orig or Dest. 

# Clean the data

### Inconsistent Column Names

```python
df_fixed = data.copy()
```

### Standarize names

```python
rename_map = {
        'oldbalanceOrg': 'oldbalance_orig',
        'newbalanceOrig': 'newbalance_orig',
        'oldbalanceDest': 'oldbalance_dest',
        'newbalanceDest': 'newbalance_dest',
        'nameOrig': 'name_orig',
        'nameDest': 'name_dest',
        'isFraud': 'is_fraud',
        'isFlaggedFraud': 'is_flagged_fraud'
    }
    
rename_map = {k: v for k, v in rename_map.items() if k in df_fixed.columns}

df_fixed = df_fixed.rename(columns=rename_map)
print(f"Renamed {len(rename_map)} columns")
for old, new in rename_map.items():
    print(f"{old} -> {new}")

else:
    print(" No columns to rename")
```

### Column isFraud with labels for Viz

```python
df_fixed['is_fraud_label'] = df_fixed['is_fraud'].map({0: 'legitimate', 1: 'fraud'})
```

```python
df_fixed.info()
```

### Fix data types

I'm going to try to reduce the size of the dataframe by analyzing the types of data and in case that there are better options, change them.


Reasoning behind the following line:
- Step to int16: it it goes from 1 to 743. I assume that the df is created monthly and we will have the first approach using this time range. In case this changes and we start having more data to work with that needs a bigger range than int16 (-32,768 to 32,767 values) we should increase the data type to int32.

- Type, nameOrig, nameDest and isFraud to category: for better memory usage.

- Amount, oldbalanceOrig, newbalanceOrig, oldbalanceDest and newbalanceDest to float32: they all go from 0.0 to a maximum of 10^8. I assume that the all the balances cannot be much more bigger (i.e. assuming business knowledge related to type of transactions) so the range of float32 (±3.4 × 10^38) would be enough.

```python
df_fixed[['step', 'is_inconsistent_origin', 'is_inconsistent_dest']] = df_fixed[['step', 'is_inconsistent_origin', 'is_inconsistent_dest']].astype('int16')
df_fixed[['type', 'name_orig', 'name_dest', 'is_fraud', 'is_flagged_fraud']] = df_fixed[['type', 'name_orig', 'name_dest', 'is_fraud', 'is_flagged_fraud']].astype('category')
df_fixed[['amount', 'oldbalance_orig', 'newbalance_orig', 'oldbalance_dest', 'newbalance_dest', 'balance_diff']] = df_fixed[['amount', 'oldbalance_orig', 'newbalance_orig', 'oldbalance_dest', 'newbalance_dest', 'balance_diff']].astype('float32')
```
